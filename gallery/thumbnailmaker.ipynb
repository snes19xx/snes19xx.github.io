```python
import os
from PIL import Image

FILE_RANGE = range(1, 27)
OUTPUT_DIR = 'thumbnails'
TARGET_QUALITY = 90
TARGET_WIDTH = 700  

# Ensure output directory exists
os.makedirs(OUTPUT_DIR, exist_ok=True)

for i in FILE_RANGE:
    filename = f"{i}.jpg"
    
    if not os.path.exists(filename):
        print(f"[MISSING] {filename}")
        continue

    try:
        with Image.open(filename) as img:
            if img.mode != 'RGB':
                img = img.convert('RGB')

            aspect_ratio = img.height / img.width
            new_height = int(TARGET_WIDTH * aspect_ratio)

            thumb = img.resize((TARGET_WIDTH, new_height), Image.Resampling.LANCZOS)

            output_path = os.path.join(OUTPUT_DIR, filename)
            thumb.save(output_path, "JPEG", quality=TARGET_QUALITY, optimize=True)
            
            print(f"[OK] {output_path} | {TARGET_WIDTH}x{new_height}px")

    except Exception as e:
        print(f"[ERROR] {filename}: {e}")
```
